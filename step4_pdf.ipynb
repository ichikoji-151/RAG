```python
from pypdf import PdfReader

reader = PdfReader("data/上級英語IB_ポイント解説 - Google スプレッドシート.pdf")
print("ページ数：", len(reader.pages))
```

```python
text = reader.pages[0].extract_text()
print(text)
```

```python
pages = []
for i ,page in enumerate(reader.pages):
    pages.append({
        "text": page.extract_text() or "",
        "source": "上級英語IB_ポイント解説 - Google スプレッドシート.pdf",
        "page" : i + 1,
    })

for p in pages[:3]:
    print(f"--- {p['source']} {p['page']}ページ ({len(['text'])}文字) ---")
    print(p["text"][:200])
```
