```python
from openai import OpenAI

client = OpenAI(
    base_url="http://127.0.0.1:1234/v1",
    api_key="lm-studio",
)

response = client.chat.completions.create(
    model="qwen3.6-35b-a3b",
    messages=[
        {"role": "system", "content": "関西弁で答えて"}, # role:systemはAIのキャラ設定、ルール
        {"role": "user", "content": "自己紹介して"}, # role:userは会話の相手の発言 
    ],
    temperature=0.2, # 0~0.3は毎回ほぼ同じ答え。0.7はバランス。1.0以降は自由奔放
)

print(response.choices[0].message.content)
```
