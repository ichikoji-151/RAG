```python
from openai import OpenAI

client = OpenAI(
    base_url="http://127.0.0.1:1234/v1",
    api_key="lm-studio",
)

# 会話履歴を入れる変数をループ外で作る
messages = [
    {"role": "system", "content" : "関西弁を答えて"}
]

while True: # ループ
    text = input("あなた> ")

    if text == "???":
        break

    if text == "exit":
        break

    # ユーザの発言を変数に追加
    messages.append({"role": "user", "content": text})

    # 変数の中身をまるごと送る
    response = client.chat.completions.create(
        model = "qwen3.6-35b-a3b",
        messages = messages,
        temperature=0.2,
    )

    answer = response.choices[0].message.content
    print("AI>", answer.strip()) # 文字列の前後にある空白や改行を取り除くメソッド

    # AIの返事の変数を追加
    messages.append({"role": "assistant", "content": answer})
    print(f"(今の履歴:{len(messages)}件)")
```
